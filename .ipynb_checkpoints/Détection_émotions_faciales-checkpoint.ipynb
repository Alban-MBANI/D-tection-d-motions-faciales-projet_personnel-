### Modèle CNN

```python
# Normalisation = Rescaling(1./48)
normalise = Rescaling(1./255)

# Augmentation modérée
data_augmentation = tf.keras.Sequential([
    RandomFlip("horizontal"),  # Sur les visages, évite "vertical" (inverser le haut/bas n'a pas de sens)
    RandomRotation(0.15),
    RandomZoom(0.15),
], name="data_augmentation")


model =Sequential([
    data_augmentation,
    normalise,

    # Bloc 1
    Conv2D(32, (3, 3), activation='relu', padding='same', input_shape=(48, 48, 1)),
    BatchNormalization(),

    Conv2D(64, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),

    # Bloc 2
    Conv2D(64, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),

    Conv2D(64, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),

    # Bloc 3
    Conv2D(128, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),

    Conv2D(128, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),

    # Bloc 4
    Conv2D(256, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),

    Conv2D(256, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),
    Dropout(0.2),  # Anti-overfitting

    # Classificateur
    Flatten(),                       # transforme les images en vecteur
    Dense(256, activation='relu'),
    BatchNormalization(),

    Dense(128, activation='relu'),
    BatchNormalization(),

    Dense(64, activation='relu'),
    BatchNormalization(),

    Dense(16, activation='relu'),
    BatchNormalization(),
    Dropout(0.5),

    Dense(1, activation='sigmoid')    # 1 neurone = binaire (happy vs sad)
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
model.build(input_shape=(None, 48, 48, 1))
model.summary()
```

```python
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)

history = model.fit(
    train_set,
    validation_data=test_set,
    epochs=30,
    callbacks=[early_stop]
)
```

### Courbe d'apprentissage

```python
loss_train, accuracy_train = model.evaluate(train_set)
loss_test, accuracy_test = model.evaluate(test_set)
print(f'Loss Train: {loss_train},Accuracy Train: {accuracy_train}')
print(f'Loss Test: {loss_test},Accuracy Test: {accuracy_test}')
```

```python
# 1. Récupération de l'historique d'entraînement
tr_acc = history.history['accuracy']
tr_loss = history.history['loss']
test_acc = history.history['val_accuracy']  # 'val_accuracy' au lieu de 'test_accuracy'
test_loss = history.history['val_loss']      # 'val_loss' au lieu de 'test_loss'

epochs = np.arange(1, len(tr_acc) + 1)

# 2. Graphique de l'Accuracy
plt.figure(figsize=(10, 6))
plt.plot(epochs, tr_acc, color='#E74C3C', linewidth=2, marker='o', label='Train Accuracy')
plt.plot(epochs, test_acc, color='#27AE60', linewidth=2, marker='s', label='Test Accuracy')
plt.title('Model Accuracy Over Epochs', fontsize=16, weight='bold')
plt.fill_between(epochs, test_acc, tr_acc, color='gray', alpha=0.1)  
plt.xlabel('Epoch', fontsize=12)
plt.ylabel('Accuracy', fontsize=12)
plt.grid(alpha=0.3)
plt.legend(fontsize=12)
plt.tight_layout()
plt.show()

# 3. Graphique de la Loss
plt.figure(figsize=(10, 6))
plt.plot(epochs, tr_loss, color='#E74C3C', linewidth=2, marker='o', label='Train Loss')
plt.plot(epochs, test_loss, color='#27AE60', linewidth=2, marker='s', label='Test Loss')
plt.title('Model Loss Over Epochs', fontsize=16, weight='bold')
plt.fill_between(epochs, test_loss, tr_loss, color='gray', alpha=0.1)  
plt.xlabel('Epoch', fontsize=12)
plt.ylabel('Loss', fontsize=12)
plt.grid(alpha=0.3)
plt.legend(fontsize=12)
plt.tight_layout()
plt.show()
```

## Matrice de Confusion du Modèle CNN

```python
# 1. Extraire à la fois les images et les vrais labels (y_true) en une seule boucle
y_true = []
y_pred_probs = []

for images, labels in test_set:
    y_true.extend(labels.numpy())
    preds = model.predict(images, verbose=0)
    y_pred_probs.extend(preds)

y_true = np.array(y_true)
y_pred_probs = np.array(y_pred_probs)

# 2. Conversion des probabilités en classes (0 ou 1) avec seuil à 0.5
y_pred_classes = (y_pred_probs > 0.5).astype(int).flatten()

# 3. Noms des classes
class_names = selected_classes  # ['happy', 'sad']

# 4. Affichage de la matrice de confusion
cm = confusion_matrix(y_true, y_pred_classes, normalize="true")

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm,
    annot=True,
    cmap="Blues",
    fmt=".2f",
    xticklabels=class_names,
    yticklabels=class_names
)
plt.xlabel("Labels Prédits", fontsize=12)
plt.ylabel("Vrais Labels", fontsize=12)
plt.title("Matrice de Confusion (Happy vs Sad)", fontsize=14)
plt.tight_layout()
plt.show()

# 5. Rapport de classification (Précision, Rappel, F1-score)
print("\nRapport de Classification :\n")
print(classification_report(y_true, y_pred_classes, target_names=class_names))
```
